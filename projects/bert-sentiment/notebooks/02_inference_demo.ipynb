# SST-2: Inference Demo

Run the fine-tuned model on a handful of sentences and inspect the
class probabilities. Requires a trained checkpoint
(`python -m sentiment.train --model <name>` from `projects/bert-sentiment/src`).

```python
from pathlib import Path

from transformers import pipeline

PROJECT_DIR = Path(__file__).resolve().parents[1]
MODEL_NAME = "bert-base"  # or "distilbert"
classifier = pipeline(
    "text-classification",
    model=str(PROJECT_DIR / "models" / MODEL_NAME),
)
```

```python
examples = [
    "The film was a fantastic, hilarious ride. I laughed the whole way."
    "Terrible acting, a boring plot, and a waste of two hours."
    "It was okay. Not great, not bad, just a movie."
    "A subtle, bittersweet story that lingers long after the credits."
]
for text in examples:
    result = classifier(text)[0]
    print(f"{result['label']:>10} ({result['score']:.4f})  {text[:60]}")
```

## Notes

- Sentences longer than 128 tokens are truncated (SST-2 has almost
  none).
- `score` is the softmax probability of the predicted label; the
  complementary probability is `1 - score`.
