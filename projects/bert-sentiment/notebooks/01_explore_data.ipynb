# SST-2: Data Exploration

Load the local SST-2 cache, inspect the class balance, and look at
sentence-length statistics and a few examples.

Run `python projects/bert-sentiment/scripts/download_data.py` from the
repository root first.

```python
from pathlib import Path

from datasets import load_from_disk

PROJECT_DIR = Path(__file__).resolve().parents[1]
raw = load_from_disk(str(PROJECT_DIR / "data" / "sst2"))
for name in ("train", "validation", "test"):
    print(name, len(raw[name]))
raw["train"].show(3)
```

```python
import numpy as np

train = raw["train"]
labels = list(train["label"])
print("negative:", labels.count(0))
print("positive:", labels.count(1))
```

```python
lengths = np.array([len(s.split()) for s in list(train["sentence"])])
print(
    f"tokens/sentence: mean={lengths.mean():.1f} "
    f"median={int(np.median(lengths))} max={lengths.max()}"
)
print(f"fraction truncated at 64 tokens: {(lengths > 64).mean():.4f}")
```

```python
idx = np.argsort(lengths)
for i in list(idx[:2]) + list(idx[-2:]):
    text = train[i]["sentence"]
    print(f"[{train[i]['label']} | {len(text.split())} tokens]")
    print(text, end="\n\n")
```
